```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(16,9))
sns.countplot(data=df, x='ai', hue='ai')
plt.title('Упоминание ИИ в вакансии')
plt.xlabel('Нет/есть ии')
plt.ylabel('Сколько раз отсутствует/встречается')
plt.legend(title='Отсутствие/наличие ии')
plt.show()

#0 - нет, 1 - есть

#подробно с countplot ознакомилась здесь: https://sky.pro/wiki/analytics/countplot-seaborn-sozdanie-i-nastrojka-schetnogo-grafika-v-python/
```

```python
df['ai'].sum()
```

Вывод: Упоминание ИИ встречается, но достаточно редко (где-то 7,2% от общего числа вакансий), подавляющее большинство без ИИ, так что, ИИ ещё не влился в большом масштабе в рынок ит, возможно, навык работы с ии настолько редок, что и зп будет выше

```python
plt.figure(figsize=(16,9))
sns.countplot(data=df, x='dms', hue='dms')
plt.title('Упоминание дмс в вакансии')
plt.xlabel('Нет/есть дмс')
plt.ylabel('Сколько раз отсутствует/встречается')
plt.legend(title='Отсутствие/наличие дмс')
plt.show()

#0 - нет, 1 - есть
```

```python
df['dms'].sum()
```

Вывод: ДМС встречается довольно часто (где-то 39%), что говорит о том, что большинство компаний предлагает дмс, но, скорее всего от этого зп становится ниже

```python
plt.figure(figsize=(16,9))
sns.countplot(data=df, x='1с', hue='1с')
plt.title('Упоминание 1c в вакансии')
plt.xlabel('Нет/есть 1c')
plt.ylabel('Сколько раз отсутствует/встречается')
plt.legend(title='Отсутствие/наличие 1c')
plt.show()

#0 - нет, 1 - есть
```

```python
df['1с'].sum()
```

Вывод: можем видеть, что 1с - относительно большая категория, но не подавляет, конечно, но количество встречи 1с достаточно внушительны, поэтому можно сказать, что спрос на автоматизацию бизнес-процессов с помощью 1с есть

```python
plt.figure(figsize=(16,9))
sns.countplot(data=df, x='education', hue='education')
plt.title('Упоминание уровня образования в вакансии')
plt.xlabel('Образование')
plt.ylabel('Сколько раз упомянуто')
plt.legend(title='Требование к уровню образования')
plt.show()

#0 - не требуется образование, 1 - требуется среднее (или может быть среднее или высшее), 2 - требуется только высшее образование
```

Вывод: в основном работадателям не важно образование, сейчас чаще смотрят на опыт, но если и смотрят, то преимущественно требуют высшее образование. Поэтому в основном рынок лоялен к отсутствию образования, но чтобы большинство дверей было открыто необходимо иметь в лучшем случае высшее образование (или хотя бы среднее, оно не откроет столько же возможностей, но больше, чем человеку без образования)

```python
plt.figure(figsize=(16,9))
sns.countplot(data=df, x='work_format', hue='work_format')
plt.title('Упоминание формата работы в вакансии')
plt.xlabel('Формат работы')
plt.ylabel('Сколько раз упомянуто')
plt.legend(title='Формат работы')
plt.show()

#0 - офис, 1 - удалёнка, 2 - гибрид
```

Вывод: офис - база, подавляющее большинство компаний предлагает работу именно из офиса, поэтому в нынешнее время, несмотря на возможности, такой формат работы выигрывает и остаётся самым востребованным, также можем обратить внимание, что второе место - гибрид, немало компаний сейчас даёт возможность сотрудникам работать на их вкус, ну и удалёнка тоже встречается, хоть и редко

```python
plt.figure(figsize=(16,9))
t = df['city'].value_counts().head(15)
t = t.reset_index()
plt.figure(figsize=(16,9))
sns.barplot(data=t, x='count', y='city', color='purple')
plt.title('Количество встречающихся городов')
plt.xlabel('Количество упоминаний')
plt.ylabel('Города')
plt.show()
```

Вывод: можем увидеть, что на графике преимущественно города-миллионники, что логично, больше людей -> больше бизнеса -> больше потребности в людях. Ну и Москва, естественно топ-1, тк весь бизнес в основе находится именно здесь + в Москве сильно развивается ИТ сфера, что требует много сотрудников, обладающих различными навыками
