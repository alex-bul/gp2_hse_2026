```python
import pandas as pd
```

```python
df = pd.read_csv('vacancies_merged_prizn (2) (1).csv', sep=',')
df
```

```python
import numpy as np
```

```python
#переиспользуем код коллег
df['payment_from_clean'] = df['payment_from']
df['payment_to_clean'] = df['payment_to']

# если payment_to меньше 1000 & payment_from > 1000, то payment_to * 1000
mask = (df['payment_to'] < 1000) & (df['payment_from'] > 1000)
df.loc[mask, 'payment_to_clean'] = df.loc[mask, 'payment_to'] * 1000

# если оба < 1000, то payment_from * 1000 + payment_to & payment_to = null
mask = (df['payment_from'] < 1000) & (df['payment_to'] < 1000)
df.loc[mask, 'payment_from_clean'] = df.loc[mask, 'payment_from'] * 1000
df.loc[mask, 'payment_to_clean'] = np.nan

# 0 = null
df.loc[df['payment_from_clean'] < 10, 'payment_from_clean'] = np.nan
df.loc[df['payment_to_clean'] < 10, 'payment_to_clean'] = np.nan
```

```python
df[df['ai'] == 1]['profession']
```

```python
df['payment_avg'] = np.where(
    (df['payment_from_clean'] > 0) & (df['payment_to_clean'] > 0),
    (df['payment_from_clean'] + df['payment_to_clean']) / 2,
    np.nan
)
```

# Гипотеза 1

H0: у вакансий связанных с ИИ зарплаты не отличаются от зарплат вакансий не связанных с ИИ

H1: у вакансий связанных с ИИ зарплаты отличаются от зарплат вакансий не связанных с ИИ

И в какую сторону и на сколько процентов они отличаются нам тоже интересно

```python
from scipy import stats
```

```python
df_temp = df.dropna(subset=['payment_avg'])
```

```python
a = df_temp['payment_avg'][df_temp['ai'] == 1]
b = df_temp['payment_avg'][df_temp['ai'] == 0]
```

```python
a.shape
```

```python
b.shape
```

Значений у нас везде > 30, по ЦПТ распределение выборочных средних сойдется к нормальному, поэтому ттест справится. Поставим доп параметр equal_var=False, чтобы не проверять неоднородность гипотез и перестраховаться


```python
import plotly.express as px
```

```python
fig = px.histogram(
    df,
    x='payment_avg',
    color='ai',
    marginal='box',
    barmode='overlay'
)

fig.show()
```

```python
stats.ttest_ind(a, b, equal_var=False)
```

```python
uplift = round(((a.mean()/b.mean()) - 1) * 100, 1)
print(f'{uplift}%')
```

pvalue < 0,05 => отклоним H0, то есть есть основания полагать, что на уровне значимости 5% пользователи в группах а и б ведут себя статзначимо различно.

То есть да, вакансии в которых есть требования к владению ИИ предлагают большую зарплату, причем большую на 62%

# Гипотеза 3
Линейная регрессия - можно ли предсказать зп по признакам - город, профессия, work format, education, dms

Что хочется узнать - насколько хорошо признаки объсняют зп (корреляция)

```python
#переиспользуем код коллег
import ast # https://habr.com/ru/companies/piter/articles/493424/

def extract_city_name(city_value):
    if pd.isna(city_value):
        return np.nan
    if str(city_value).startswith('{'):
          city_dict = ast.literal_eval(city_value)
          return city_dict.get('title', np.nan)
    else:
          return city_value

df['city_name'] = df['city'].apply(extract_city_name)
```

```python
#переиспользуем код коллег
city_counts = df['city_name'].value_counts()

min_count = 200
top_cities = city_counts[city_counts >= min_count].index.tolist() #убираем те, которые встречаются меньше 200 раз

df['city_grouped'] = df['city_name'].apply(
    lambda x: x if x in top_cities else 'Другие'
)

df['city_grouped'].nunique()
```

```python
df = pd.get_dummies(df, columns=['city_grouped'], drop_first=True)
```

```python
city_cols = [i for i in df.columns if i.startswith('city_grouped_')]
```

```python
c = ['profession', 'work_format', 'education', 'dms', 'payment_avg', 'ai']
df_clean = df[c + city_cols].dropna(subset=c)

top_15 = df_clean['profession'].value_counts().nlargest(15).index
df_clean['profession_grouped'] = df_clean['profession'].apply(
    lambda x: x if x in top_15 else 'Другие')
```

Построим линейную регрессию методом OLS (метод наименьших квадратов), чтобы вместо ручного подсчете R2, pvalue и коэффициентов линейной регресии, можно было автоматически получить влияние (коэффициенты) каждого признака
(профессии, города, образования) на целевую переменную - зарплату, и еще и проверку их статзначимости.

```python
import statsmodels.api as sm

X = pd.get_dummies(df_clean[['profession_grouped', 'work_format']], drop_first=True)

X = pd.concat([X, df_clean[city_cols[1:]]], axis=1)

X['education'] = df_clean['education']
X['dms'] = df_clean['dms']
X['ai'] = df_clean['ai']

y = df_clean['payment_avg']

X = sm.add_constant(X).astype(float)
model = sm.OLS(y, X).fit()

print(model.summary())
```

Выводы:  
1) R2 = 0.28
наши признаки (город, профессия, ии, наличие дмс, уровень образования) объясняют 28% разброса зарплат -> то есть все таки есть какая-то статзначимая свзяь между зп и признаками, hr не ставят зп на рандом
2) У Москвы самый большой статзначимый вес из городов - то есть действительно в Москве выше зп
3) Из профессий самая большая связь у разработчика

Еще раз подтвердили статзначимое влияние ии, образования и дмс на зп
