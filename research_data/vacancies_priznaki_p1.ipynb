```python
import pandas as pd
```

```python
df = pd.read_csv('vacancies_merged.csv')
```

```python
import re
```

```python
if 'dms' in df.columns:
    df = df.drop('dms', axis=1)
slow = r'дмс|медицинск.*страхов.*'
df['dms'] = 0
for ind in df.index:
  opis = df['vacancy_description'].iloc[ind]
  if pd.notna(opis):
    if re.search(slow, opis, re.IGNORECASE):
      df.loc[ind, 'dms'] = 1
df['dms'].sum()

#https://sky.pro/wiki/python/osnovnye-funkcii-i-metody-modulya-re/ регулярные выражениизучила тут
```

```python
df
```

```python
if 'ai' in df.columns:
    df = df.drop('ai', axis=1)
slow = r'\bai\b|\bai(?=-)|\bai(?= )|\bии\b|искусственн.*интеллект.*|\bml\b|машинно.*обучен.*|генеративн.*модел.*|\bllm\b|нейросет.*|\bdl\b|deep learning|\bnlp\b|generative ai|computer vision|нейронны.*сет.*'
df['ai'] = 0
for ind in df.index:
  opis = df['vacancy_description'].iloc[ind]
  if pd.notna(opis):
    if re.search(slow, opis, re.IGNORECASE):
      df.loc[ind, 'ai'] = 1
df['ai'].sum()

#предположила что есть вариант что написано что-то типо ai-... или ai ..., поэтому обратилась к статье:
#https://habr.com/ru/articles/956068/
```

```python
df
```

```python
if 'work_format' in df.columns:
    df = df.drop('work_format', axis=1)
slowg = r'гибри.*|частично удал.*|смешанный вариант|смешанный режим работы|смешанный формат'
slowr = r'удален.*|удалён.*|remote work|работа из дома'
df['work_format'] = 0
for ind in df.index:
  opis = df['vacancy_description'].iloc[ind]
  if pd.notna(opis):
    if re.search(slowr, opis, re.IGNORECASE):
      df.loc[ind, 'work_format'] = 2
    elif re.search(slowg, opis, re.IGNORECASE):
      df.loc[ind, 'work_format'] = 1
```

```python
df
```

```python
if 'education' in df.columns:
    df = df.drop('education', axis=1)
v = r'высшее'
sr = r'среднее'
df['education'] = 0
for ind in df.index:
  opis = df['vacancy_description'].iloc[ind]
  if pd.notna(opis):
    if re.search(sr, opis, re.IGNORECASE):
      df.loc[ind, 'education'] = 1
    elif re.search(v, opis, re.IGNORECASE):
      df.loc[ind, 'education'] = 2
```

```python
df
```

```python
if '1с' in df.columns:
    df = df.drop('1с', axis=1)
c = r'\b1с.*|\b1c.*'
df['1с'] = 0
for ind in df.index:
  opis = df['vacancy_description'].iloc[ind]
  if pd.notna(opis):
    if re.search(c, opis, re.IGNORECASE):
      df.loc[ind, '1с'] = 1
df['1с'].sum()
```
